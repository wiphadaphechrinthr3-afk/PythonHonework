4. เมื่อทำครบทุกตอน ให้รันเซลล์ **สรุปผลการประเมิน** ท้ายแบบฝึกหัด

> ⏱ เวลาที่แนะนำ: 60–90 นาที

````python
#@title ▶ รันเซลล์นี้ก่อนเริ่มทำแบบฝึกหัด (ระบบตรวจคำตอบอัตโนมัติ — ไม่ต้องแก้ไข)
import hashlib, io, contextlib, subprocess, sys, keyword

_SALT = "py3-ch4"
_KEY = {"A1": "c6448ce783c35268", "A2": "9c66c94d608657aa", "A3": "743a89d55d7af7f1", "A4": "d07789e379eadd8f", "A5": "90cbe8d57919345e", "A6": "b3654a031d208b35", "A7": "481de27bb8690348", "A8": "a3ea7f554507e510", "A9": "dfb8ceb46ccd7b37", "A10": "221629cc3b326101", "A11": "9b0c15bb9ab2f8d1", "A12": "7779022dc4ac046d", "A13": "e0491a62bc04de02", "A14": "3b54e42fc2124aa6", "A15": "2f1705bdf8e0a720", "A16": "236484640f57ff1a", "A17": "e8edd9959f39e6d7", "A18": "816b99928019eec5", "A19": "7d64757a5661eb83", "A20": "d04a2be4fec8f580", "B1": "0ac0ad4d1e024241", "B2": "fa61dd7619f519d6", "B3": "c5530fc66aed6afb", "B4": "7d4766e06a550b05", "B5": "04836c0488b44acc", "B6": "4575d5068c677535", "B7": "e4c8571ff61a62d5", "B8": "9138d93a4b0c14c5", "B9": "da5d228bb874c5bb", "B10": "bfe413e57a94d3f2", "C1": "8ec028ebdd624aa1", "C2": "8e5d7d835ada3be4", "C3": "a806f9a98ec68dee", "C4": "64bdb2dc4520ac29", "C5": "7170860a0efba7b0", "C6": "939f2ff9d1a1f03c", "C7": "40c1a1c0e6493272", "C8": "f082c7e1d96f6e17", "E2": "cfe54d63850b2e63", "E3": "5b626814070c6e0a", "E4": "27f1274fadd65693", "E5": "1d993189796f1bf0", "E6": "4682b84747865ae5", "E7": "58246f18df382622", "E8": "0cb3b7edca007140", "E10": "34f8f137d35ca1bf"}
_TOPIC = {"A1": "4.1", "A2": "4.1.1", "A3": "4.1.2", "A4": "4.1.1", "A5": "4.1.2", "A6": "4.1.3", "A7": "4.1.4", "A8": "4.1.4", "A9": "4.1.5", "A10": "4.1.6", "A11": "4.1.8", "A12": "4.2.1", "A13": "4.2.2", "A14": "4.2.2", "A15": "4.2.2", "A16": "4.2.2", "A17": "4.3.3", "A18": "4.3.2", "A19": "4.3.2", "A20": "4.4", "B1": "4.1.3", "B2": "4.1.3", "B3": "4.1.1", "B4": "4.1.7", "B5": "4.1.5", "B6": "4.1.6", "B7": "4.2.1", "B8": "4.2.2", "B9": "4.2.2", "B10": "4.3.3", "C1": "4.2.2", "C2": "4.2.2", "C3": "4.2.2", "C4": "4.2.2", "C5": "4.2.2", "C6": "4.2.2", "C7": "4.1.4, 4.1.6", "C8": "4.1.8, 4.2.2", "E2": "4.2.2", "E3": "4.2.2", "E4": "4.2.2", "E5": "4.2.2", "E6": "4.2.2", "E7": "4.1.4, 4.1.5", "E8": "4.1.8", "E9": "4.1.2", "E10": "4.3.3"}
_PARTS = [('ตอนที่ 1 ปรนัย', ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10', 'A11', 'A12', 'A13', 'A14', 'A15', 'A16', 'A17', 'A18', 'A19', 'A20']), ('ตอนที่ 2 ถูก/ผิด', ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10']), ('ตอนที่ 3 ทำนายผลลัพธ์', ['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8']), ('ตอนที่ 5 เขียนโปรแกรม', ['E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10'])]
_SNIP = {"C1": "print(\"Sammy has {} balloons and {} kites.\".format(3, 7))", "C2": "print(\"{2}-{0}-{1}\".format(\"a\", \"b\", \"c\"))", "C3": "print(\"{name} is {age} years old.\".format(age=20, name=\"Somchai\"))", "C4": "print(\"Price: {0:.1f} baht ({1})\".format(49.95, \"sale\"))", "C5": "print(\"[{:>6}]\".format(\"Hi\"))\nprint(\"[{:<6}]\".format(\"Hi\"))", "C6": "print(\"{:-^11}\".format(\"Python\"))", "C7": "total = 1 + \\\n        2 + \\\n        3\nprint(total)  # แสดงผลรวม", "C8": "x = \"ab\"; y = \"-\"; print(\"{0}{1}{0}\".format(x, y))"}
_results = {}

def _h(qid, value):
    return hashlib.sha256(f"{_SALT}|{qid}|{value}".encode("utf-8")).hexdigest()[:16]

def _norm_text(s):
    lines = [ln.rstrip() for ln in str(s).replace("\r\n", "\n").split("\n")]
    while lines and lines[0] == "":
        lines.pop(0)
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _norm_choice(prefix, ans):
    a = str(ans).strip().upper()
    if prefix == "B":
        if a in ("T", "TRUE", "ถูก", "ท", "1"): return "T"
        if a in ("F", "FALSE", "ผิด", "ผ", "0"): return "F"
    return a

def grade(part, answers):
    """ตรวจคำตอบตอนที่ 1 (part='A') และตอนที่ 2 (part='B')"""
    n_ok, n_all = 0, 0
    for n in sorted(answers):
        qid = f"{part}{n}"
        if qid not in _KEY:
            continue
        n_all += 1
        a = _norm_choice(part, answers[n])
        if a == "":
            print(f"ข้อ {n:>2}: ⚪ ยังไม่ได้ตอบ")
            _results[qid] = False
            continue
        ok = _h(qid, a) == _KEY[qid]
        _results[qid] = ok
        n_ok += ok
        print(f"ข้อ {n:>2}: {'✅ ถูกต้อง' if ok else '❌ ยังไม่ถูก  (ทบทวนหัวข้อ ' + _TOPIC[qid] + ')'}")
    print("-" * 40)
    print(f"ได้ {n_ok} / {n_all} คะแนน")

def predict(qid, prediction):
    """ตอนที่ 3: เปรียบเทียบผลลัพธ์ที่ทำนายกับผลลัพธ์จริง (บันทึกคะแนนจากการตรวจครั้งแรก)"""
    pred = _norm_text(prediction)
    if pred == "":
        print("⚪ ยังไม่ได้เขียนคำตอบที่ทำนายไว้ในตัวแปร")
        return
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(_SNIP[qid], {})
    actual = _norm_text(buf.getvalue())
    ok = _h(qid, pred) == _KEY[qid]
    first = qid not in _results
    if first:
        _results[qid] = ok
    print("✅ ทำนายถูกต้อง" if ok else "❌ ทำนายไม่ตรงกับผลลัพธ์จริง")
    print("\nผลลัพธ์ที่ทำนาย:")
    print(pred)
    print("\nผลลัพธ์จริงจากการรันโค้ด:")
    print(actual)
    if not ok:
        print("\n(สังเกตช่องว่าง เครื่องหมาย และลำดับของค่าให้ดี — ทบทวนหัวข้อ " + _TOPIC[qid] + ")")
    if not first:
        print("\nℹ️ คะแนนข้อนี้บันทึกจากการตรวจครั้งแรกไปแล้ว")

def check(qid, value):
    """ตอนที่ 5: ตรวจค่าที่ได้จากโปรแกรม"""
    ok = _h(qid, repr(value)) == _KEY[qid]
    _results[qid] = ok
    if ok:
        print("✅ ถูกต้อง เยี่ยมมาก!")
    else:
        print("❌ ยังไม่ถูกต้อง ค่าที่ได้คือ:")
        print(repr(value))
        print("ตรวจสอบความกว้างฟิลด์ การจัดตำแหน่ง จำนวนทศนิยม และช่องว่างอีกครั้ง (หัวข้อ " + _TOPIC[qid] + ")")

def check_keywords(n_keywords, found):
    """ตอนที่ 5 ข้อ E9: ตรวจจำนวนคำสงวนและคำที่เป็นคำสงวน"""
    words = ["if", "If", "print", "class", "name", "True", "lambda", "input", "pass", "Else"]
    expect = [w for w in words if keyword.iskeyword(w)]
    ok1 = n_keywords == len(keyword.kwlist)
    ok2 = list(found) == expect
    _results["E9"] = ok1 and ok2
    print(("✅" if ok1 else "❌") + " จำนวนคำสงวน")
    print(("✅" if ok2 else "❌") + " รายการคำที่เป็นคำสงวน")

def check_script(filename):
    """ตอนที่ 5 ข้อ E10: รันไฟล์สคริปต์ในโหมดสคริปต์แล้วตรวจผลลัพธ์"""
    try:
        r = subprocess.run([sys.executable, filename], capture_output=True, text=True, timeout=20)
    except FileNotFoundError:
        print("❌ ไม่พบไฟล์", filename, "— รันเซลล์ %%writefile ก่อน")
        _results["E10"] = False
        return
    out = _norm_text(r.stdout)
    print("ผลลัพธ์จากการรัน python", filename)
    print(out if out else "(ไม่มีผลลัพธ์)")
    if r.stderr:
        print("ข้อผิดพลาด:\n" + r.stderr)
    ok = _h("E10", out) == _KEY["E10"]
    _results["E10"] = ok
    print("\n✅ ถูกต้อง" if ok else "\n❌ ผลลัพธ์ยังไม่ตรงตามที่โจทย์กำหนด")

def summary():
    """สรุปผลการประเมินตนเอง"""
    total_ok, total_all, review = 0, 0, []
    print(f"{'ส่วน':<28}{'คะแนน':>10}")
    print("=" * 38)
    for name, qids in _PARTS:
        ok = sum(1 for q in qids if _results.get(q))
        total_ok += ok
        total_all += len(qids)
        print(f"{name:<28}{ok:>5} / {len(qids):<3}")
        review += [q for q in qids if not _results.get(q)]
    print("=" * 38)
    pct = 100 * total_ok / total_all
    print(f"{'รวม (ส่วนที่ตรวจอัตโนมัติ)':<28}{total_ok:>5} / {total_all:<3}  ({pct:.1f}%)")
    print()
    if pct >= 80:
        print("🌟 ดีมาก — เข้าใจเนื้อหาบทที่ 4 เป็นอย่างดี พร้อมเรียนบทถัดไป")
    elif pct >= 60:
        print("👍 ดี — เข้าใจเนื้อหาส่วนใหญ่ ควรทบทวนหัวข้อที่ยังทำไม่ได้")
    else:
        print("📖 ควรทบทวนบทเรียนอีกครั้ง แล้วกลับมาทำแบบฝึกหัดใหม่")
    topics = sorted({t.strip() for q in review for t in _TOPIC[q].split(",")},
                    key=lambda s: [int(x) for x in s.split(".")])
    if topics:
        print("\nหัวข้อที่ควรทบทวน:", ", ".join(topics))
    print("\n* ตอนที่ 4 และตอนที่ 6 ให้ตรวจสอบกับผลลัพธ์ที่โจทย์กำหนด/ผู้สอน")

print("✅ ระบบตรวจคำตอบพร้อมใช้งาน — เริ่มทำแบบฝึกหัดได้เลย")
````
